# CoinGecko — Exploração
Notebook para análise ad-hoc. Roda localmente via Databricks Connect.

```python
from databricks.connect import DatabricksSession

spark = DatabricksSession.builder.serverless().profile("My Free Edition").getOrCreate()
print("Conectado:", spark.version)
```

```python
# Ranking atual de moedas
spark.sql("""
  SELECT *
  FROM coingecko.gold.gold_ranking_moedas_diario
  WHERE data_referencia = (SELECT MAX(data_referencia) FROM coingecko.gold.gold_ranking_moedas_diario)
  ORDER BY rank_valor_mercado
""").toPandas()
```

```python
# Resumo do mercado — últimos 30 dias
spark.sql("""
  SELECT
    data_referencia,
    qtd_moedas,
    ROUND(valor_mercado_total_usd / 1e12, 2) AS market_cap_tri,
    ROUND(volume_total_usd / 1e9, 2)         AS volume_bi,
    top_valor_mercado_nome,
    top_volume_nome
  FROM coingecko.gold.gold_resumo_mercado_diario
  ORDER BY data_referencia DESC
  LIMIT 30
""").toPandas()
```

```python
# Variação D1 — maiores altas e baixas do último dia
spark.sql("""
  SELECT id_moeda, simbolo, nome_moeda, preco_medio_usd, variacao_pct_d1
  FROM coingecko.gold.gold_variacao_preco_d1
  WHERE data_referencia = (SELECT MAX(data_referencia) FROM coingecko.gold.gold_variacao_preco_d1)
  ORDER BY variacao_pct_d1 DESC
""").toPandas()
```
